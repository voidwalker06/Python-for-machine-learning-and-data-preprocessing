# Week 1 Project: Python for Machine Learning & Data Preprocessing

**Project:** Student Performance Dataset Cleaning and Preprocessing  
**Tools:** Python, NumPy, Pandas, Matplotlib, and Scikit-learn

## Aim
To load a sample dataset, explore it, clean missing and duplicate values, select useful features, encode categorical data, normalize numerical data, and save a prepared dataset.

The dataset is synthetic and created only for learning and demonstration.

## 1. Import libraries and load the dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from pathlib import Path

data_file = Path("student_performance_raw.csv")
df = pd.read_csv(data_file)
df.head()
```

## 2. Inspect the data and perform exploratory data analysis (EDA)

```python
print("Rows and columns:", df.shape)
print("\\nColumn information:")
df.info()
print("\\nDescriptive statistics:")
display(df.describe(include="all"))
print("\\nMissing values per column:")
display(df.isnull().sum())
print("\\nDuplicate rows:", df.duplicated().sum())
```

```python
df["Final_Score"].plot(kind="hist", bins=8, edgecolor="black",
                       title="Distribution of Final Scores")
plt.xlabel("Final score")
plt.ylabel("Number of students")
plt.show()
```

## 3. Remove duplicate rows and handle missing values

- Duplicate rows are removed.
- Missing numerical values in `Study_Hours` and `Attendance_Percent` are replaced with their respective medians.
- If a categorical column contains missing values, its mode is used.

Median is a useful choice for numerical data because it is less affected by unusually high or low values.

```python
df = df.drop_duplicates()
df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].median())
df["Attendance_Percent"] = df["Attendance_Percent"].fillna(df["Attendance_Percent"].median())

for column in ["Gender", "Area", "Internet_Access"]:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])

print(df.isnull().sum())
```

## 4. Feature selection

`Student_ID` is removed because it is only an identifier. `Final_Score` is the target variable. The remaining columns are used as input features.

```python
selected_features = [
    "Age", "Gender", "Area", "Study_Hours",
    "Attendance_Percent", "Internet_Access"
]
X = df[selected_features].copy()
y = df["Final_Score"].copy()
X.head()
```

## 5. Encode categorical variables

```python
X_encoded = pd.get_dummies(
    X,
    columns=["Gender", "Area", "Internet_Access"],
    drop_first=True,
    dtype=int
)
X_encoded.head()
```

## 6. Normalize numerical features

```python
numeric_columns = ["Age", "Study_Hours", "Attendance_Percent"]
scaler = MinMaxScaler()
X_encoded[numeric_columns] = scaler.fit_transform(X_encoded[numeric_columns])
X_encoded[numeric_columns].describe()
```

## 7. Save the cleaned dataset

```python
processed = X_encoded.copy()
processed["Final_Score"] = y.values
processed.to_csv("student_performance_cleaned.csv", index=False)
processed.head()
```

## 8. Conclusion

In this project, Python libraries were used to load and inspect a sample dataset. Missing values were handled, duplicate records were removed, useful features were selected, categorical columns were encoded, and numerical features were normalized. Exploratory data analysis was also performed using summary statistics and a histogram. The final cleaned dataset was saved as a CSV file and is ready for use in a later machine learning workflow.
